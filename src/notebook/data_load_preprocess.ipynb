```python
#Data Load and Initial EDA analysis part Week5 Delivarbles
# Note : Code required intractive google drvie mount to get CTK image data (40 GB)
# --- SETUP --------------------------------------------------------------------
# Colab + Core libs
!pip install pydicom imageio
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from glob import glob
import os, re, math, textwrap, itertools, statistics
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Stats & ML utilities
from scipy import stats
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from collections import Counter, defaultdict

# Imaging
import pydicom
import imageio.v2 as iio

# --- PROJECT PATHS (edit if your layout differs) ------------------------------
PROJECT_ROOT = Path('/content/drive/My Drive/capstone/tumor_pred_ai')
DATASET_ROOT = PROJECT_ROOT / 'DataSet'
CANDIDATE_RAW = [DATASET_ROOT/'manifest-1592488683281', DATASET_ROOT/'raw', DATASET_ROOT/'Raw']

RAW_DIR = next((p for p in CANDIDATE_RAW if p.exists()), None)
if RAW_DIR is None:
    matches = list(DATASET_ROOT.rglob('C4KC-KiTS_Clinical-Data_Version.csv'))
    if matches: RAW_DIR = matches[0].parent
    else:
        raise FileNotFoundError(f"Could not find RawData-S or the clinical CSV under {DATASET_ROOT}")

CLINICAL_CSV = RAW_DIR / 'C4KC-KiTS_Clinical-Data_Version.csv'
META_CSV     = RAW_DIR / 'metadata.csv'
IMAGE_ROOT   = RAW_DIR / 'C4KC-KiTS'

print("Using RAW_DIR:", RAW_DIR)
print("Clinical CSV:", CLINICAL_CSV)
print("Metadata CSV:", META_CSV)

# --- LOAD DATA ----------------------------------------------------------------
assert CLINICAL_CSV.exists(), f"Missing: {CLINICAL_CSV}"
assert META_CSV.exists(), f"Missing: {META_CSV}"

# Meaningful dataset names (avoid df1, v1, etc.)
clinical_df = pd.read_csv(CLINICAL_CSV)     # example name: rows x cols
meta_df = pd.read_csv(META_CSV)

# Discover images (DICOM preferred)
if not IMAGE_ROOT.exists():
    # fallback: pick subdir with most image-like files
    candidates = [p for p in RAW_DIR.iterdir() if p.is_dir()]
    best, best_cnt = RAW_DIR, -1
    for d in candidates:
        cnt = sum(len(glob(str(d/'**'/'*.'+ext), recursive=True)) for ext in ['dcm','png','jpg','jpeg'])
        if cnt > best_cnt: best, best_cnt = d, cnt
    IMAGE_ROOT = best

dicoms = glob(str(IMAGE_ROOT/'**'/'*.dcm'), recursive=True)
pngs   = glob(str(IMAGE_ROOT/'**'/'*.png'), recursive=True)
jpgs   = glob(str(IMAGE_ROOT/'**'/'*.jpg'), recursive=True) + glob(str(IMAGE_ROOT/'**'/'*.jpeg'), recursive=True)

print(f"Image root: {IMAGE_ROOT}")
print(f"Found {len(dicoms)} DICOM, {len(pngs)} PNG, {len(jpgs)} JPG")

# --- QUICK PREVIEW (DICOM if available) ---------------------------------------
def normalize_to_2d(arr):
    """
    Ensure the returned image is 2D for imshow().
    - If (H,W): return as-is
    - If (F,H,W): take middle frame
    - If (H,W,3/4): keep first 3 channels or convert to grayscale
    """
    if arr.ndim == 2:
        return arr
    if arr.ndim == 3 and arr.shape[-1] in (3,4):
        return arr[..., :3]  # RGB/RGBA
    if arr.ndim == 3:
        return arr[arr.shape[0]//2]
    raise TypeError(f"Unexpected image shape {arr.shape}")

def show_image(path):
    plt.figure(figsize=(4,4))
    try:
        if path.lower().endswith('.dcm'):
            ds = pydicom.dcmread(path)
            img = normalize_to_2d(ds.pixel_array)
            plt.imshow(img, cmap='gray'); plt.title(os.path.basename(path)); plt.axis('off')
        else:
            img = iio.imread(path)
            img = normalize_to_2d(np.asarray(img))
            plt.imshow(img, cmap='gray'); plt.title(os.path.basename(path)); plt.axis('off')
    except Exception as e:
        print("Preview failed:", e)
    plt.show()

sample_img = dicoms[0] if dicoms else (pngs[0] if pngs else (jpgs[0] if jpgs else None))
if sample_img:
    print("Showing sample image:", sample_img)
    show_image(sample_img)
else:
    print("No images found to preview.")

# --- DATA DICTIONARY ----------------------------------------------------------
def infer_semantic_type(series: pd.Series):
    """Heuristic mapping to nominal/ordinal/binary/numeric/date/bool."""
    if pd.api.types.is_bool_dtype(series): return 'binary' # Handle boolean type directly
    s = series.dropna()
    if s.empty: return 'unknown'
    # boolean-like?
    low = set(map(lambda x: str(x).strip().lower(), s.unique().tolist()))
    if low.issubset({'true','false','0','1','yes','no'}): return 'binary'
    # datetime?
    if np.issubdtype(series.dtype, np.datetime64): return 'date'
    # numeric?
    if pd.api.types.is_numeric_dtype(series): return 'numeric'
    # short cardinality -> nominal
    if s.nunique() <= max(10, len(s)//5): return 'nominal'
    return 'text'

def build_dictionary(df: pd.DataFrame, dataset_name: str):
    rows = []
    for col in df.columns:
        ser = df[col]
        # attempt safe type cast for dates/bools without mutating original df
        ser_cast = ser.copy()
        if ser.dtype == object:
            # try datetime
            try:
                ser_cast_dt = pd.to_datetime(ser_cast, errors='raise')
                ser_cast = ser_cast_dt
            except Exception:
                pass
        sem_type = infer_semantic_type(ser_cast)
        ex_vals = ser.dropna().astype(str).unique()[:3]
        ex_vals = ', '.join(map(lambda s: (s[:30]+'…') if len(s)>30 else s, ex_vals))
        rows.append({
            'dataset': dataset_name,
            'column': col,
            'pandas_dtype': str(ser.dtype),
            'semantic_type': sem_type,
            'non_null': ser.notna().sum(),
            'nulls': ser.isna().sum(),
            '%null': round(100*ser.isna().mean(), 2),
            'n_unique': ser.nunique(dropna=True),
            'example_values': ex_vals
        })
    return pd.DataFrame(rows).sort_values(['semantic_type','column']).reset_index(drop=True)

dict_clinical = build_dictionary(clinical_df, 'clinical')
dict_metadata = build_dictionary(meta_df, 'metadata')

print("Data Dictionary — clinical (top 15 rows)")
display(dict_clinical.head(15))
print("Data Dictionary — metadata (top 15 rows)")
display(dict_metadata.head(15))

# Save for documentation
out_dir = PROJECT_ROOT / 'artifacts'
out_dir.mkdir(parents=True, exist_ok=True)
dict_clinical.to_csv(out_dir/'data_dictionary_clinical.csv', index=False)
dict_metadata.to_csv(out_dir/'data_dictionary_metadata.csv', index=False)
print("Saved data dictionaries to:", out_dir)

# --- TYPE NORMALIZATION / PARSING ---------------------------------------------
clinical_df = clinical_df.copy()
meta_df = meta_df.copy()

# Example conversions:
# 1) Convert obvious boolean-like columns in clinical
bool_like_cols = [c for c in clinical_df.columns
                  if clinical_df[c].dropna().astype(str).str.lower().isin(['true','false']).all()]
for c in bool_like_cols:
    clinical_df[c] = clinical_df[c].map({'True': True, 'False': False, 'true': True, 'false': False})

# 2) Convert numeric-looking columns stored as strings
for c in clinical_df.columns:
    if clinical_df[c].dtype == object:
        try:
            clinical_df[c] = pd.to_numeric(clinical_df[c], errors='ignore')
        except Exception:
            pass

# 3) Parse dates in metadata
for dc in ['Study Date', 'Download Timestamp']:
    if dc in meta_df.columns:
        meta_df[dc] = pd.to_datetime(meta_df[dc], errors='coerce')

# 4) Light corpus parsing on text fields (if applicable)
text_fields = [c for c in meta_df.columns if meta_df[c].dtype == object]
def top_tokens(series, k=15):
    tokens = []
    for txt in series.dropna().astype(str):
        toks = re.findall(r"[A-Za-z0-9\-]+", txt.lower())
        tokens.extend(toks)
    return Counter(tokens).most_common(k)

for tf in text_fields:
    tok = top_tokens(meta_df[tf], k=10)
    if tok:
        print(f"Top tokens in '{tf}':", tok)

# --- DEDUPLICATION -------------------------------------------------------------
def report_duplicates(df, key_cols=None, name='df'):
    print(f"\nDuplicate check for {name}:")
    # full-row duplicates
    dup_full = df.duplicated().sum()
    print("  Full-row duplicates:", dup_full)
    if key_cols:
        for k in key_cols:
            if k in df.columns:
                dup_k = df[k].duplicated().sum()
                print(f"  Duplicates by '{k}':", dup_k)

report_duplicates(clinical_df, key_cols=['patient_id'], name='clinical')
report_duplicates(meta_df, key_cols=['Series UID','Subject ID','Study UID'], name='metadata')

# Drop exact duplicates if any
clinical_df = clinical_df.drop_duplicates().reset_index(drop=True)
meta_df = meta_df.drop_duplicates().reset_index(drop=True)

# --- MISSING / INCONSISTENCIES ------------------------------------------------
def missing_report(df, name):
    miss = df.isna().mean().sort_values(ascending=False)
    print(f"\nMissingness — {name} (top 10)")
    display((miss*100).round(2).head(10).to_frame('%missing'))
missing_report(clinical_df, 'clinical')
missing_report(meta_df, 'metadata')

# Example simple inconsistency: negative ages or impossible BMI
if 'age_at_nephrectomy' in clinical_df.columns:
    bad_age = clinical_df['age_at_nephrectomy'][(clinical_df['age_at_nephrectomy']<0)|(clinical_df['age_at_nephrectomy']>120)]
    print("Implausible ages:", bad_age.tolist())
if 'body_mass_index' in clinical_df.columns:
    bad_bmi = clinical_df['body_mass_index'][(clinical_df['body_mass_index']<10)|(clinical_df['body_mass_index']>90)]
    print("Implausible BMI:", bad_bmi.tolist())

# --- SKEWNESS / TYPE CONVERSIONS SUMMARY --------------------------------------
def numeric_cols(df):
    return [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
num_clin = numeric_cols(clinical_df)
num_meta = numeric_cols(meta_df)

def skew_table(df, cols, name):
    recs=[]
    for c in cols:
        s = df[c].dropna()
        if len(s)>1:
            recs.append({'column': c, 'n': len(s), 'mean': s.mean(), 'std': s.std(), 'skew': stats.skew(s)})
    tab = pd.DataFrame(recs).sort_values('skew', key=lambda x: x.abs(), ascending=False)
    print(f"\nSkewness — {name} (top 10 by |skew|)")
    display(tab.head(10))
skew_table(clinical_df, num_clin, 'clinical')
skew_table(meta_df, num_meta, 'metadata')

# --- OUTLIERS (z-score>3) -----------------------------------------------------
def outliers_by_z(df, cols, name, z=3.0):
    outs = {}
    for c in cols:
        s = df[c]
        if s.notna().sum() > 3:
            zscores = np.abs(stats.zscore(s.dropna()))
            idx = s.dropna().index[zscores>z]
            if len(idx)>0:
                outs[c] = df.loc[idx, c]
    print(f"\nOutliers (|z|>{z}) — {name}")
    for c, ser in outs.items():
        print(f"  {c}: {len(ser)} outlier(s) -> {ser.values[:5]}")
outliers_by_z(clinical_df, num_clin, 'clinical', z=3.0)
outliers_by_z(meta_df, num_meta, 'metadata', z=3.0)

# --- IMBALANCE (categoricals) -------------------------------------------------
def plot_bar(series, title):
    counts = series.value_counts(dropna=False)
    plt.figure(figsize=(6,4))
    counts.plot(kind='bar')  # single plot; no explicit colors
    plt.title(title); plt.ylabel('count'); plt.xlabel('class')
    plt.xticks(rotation=30, ha='right')
    plt.tight_layout(); plt.show()

for cat in ['gender','vital_status','Modality','Series Description']:
    if cat in clinical_df.columns:
        plot_bar(clinical_df[cat], f"Class distribution: {cat} (clinical)")
    elif cat in meta_df.columns:
        plot_bar(meta_df[cat], f"Class distribution: {cat} (metadata)")

# --- CORRELATIONS & LINEARITY -------------------------------------------------
def corr_heatmap(df, title):
    cols = numeric_cols(df)
    if len(cols) < 2:
        print(f"Not enough numeric columns for correlation: {title}")
        return None # Return None if no correlation heatmap is generated
    C = df[cols].corr(numeric_only=True)
    plt.figure(figsize=(6,5))
    plt.imshow(C, interpolation='nearest')
    plt.colorbar()
    plt.title(title)
    plt.xticks(range(len(cols)), cols, rotation=90)
    plt.yticks(range(len(cols)), cols)
    plt.tight_layout(); plt.show()
    return C

C_clin = corr_heatmap(clinical_df, "Correlation — clinical")
C_meta = corr_heatmap(meta_df, "Correlation — metadata")

# --- NORMALIZATION / STANDARDIZATION ------------------------------------------
scaler_std = StandardScaler()
scaler_mm  = MinMaxScaler()

def scale_numeric(df, prefix):
    cols = numeric_cols(df)
    if not cols: return df.copy(), []
    X = df[cols].astype(float).fillna(df[cols].astype(float).median())
    std_scaled = pd.DataFrame(scaler_std.fit_transform(X), columns=[f"{c}_z" for c in cols])
    mm_scaled  = pd.DataFrame(scaler_mm.fit_transform(X),  columns=[f"{c}_minmax" for c in cols])
    out = pd.concat([df.reset_index(drop=True), std_scaled, mm_scaled], axis=1)
    return out, cols

clinical_scaled, clin_num_cols = scale_numeric(clinical_df, 'clinical')
metadata_scaled, meta_num_cols = scale_numeric(meta_df, 'metadata')
print("Scaled clinical shape:", clinical_scaled.shape)
print("Scaled metadata shape:", metadata_scaled.shape)

# --- SIMPLE IMAGE DATA AUGMENTATION (on one DICOM slice) ----------------------
# Note: For medical images, prefer intensity-preserving + small geometric changes.
def random_flip(img):
    if np.random.rand()<0.5: img = np.flipud(img)
    if np.random.rand()<0.5: img = np.fliplr(img)
    return img

def random_rotate_90(img):
    k = np.random.choice([0,1,2,3])
    return np.rot90(img, k)

def random_gaussian_noise(img, sigma=5.0):
    return np.clip(img + np.random.normal(0, sigma, img.shape), 0, np.max(img))

def show_augments(dcm_path):
    ds = pydicom.dcmread(dcm_path)
    base = normalize_to_2d(ds.pixel_array.astype(np.float32))
    aug1 = random_flip(base.copy())
    aug2 = random_rotate_90(base.copy())
    aug3 = random_gaussian_noise(base.copy(), sigma=5.0)

    for title, im in [("original", base), ("flip", aug1), ("rotate90", aug2), ("gaussian_noise", aug3)]:
        plt.figure(figsize=(4,4)); plt.imshow(im, cmap='gray'); plt.title(title); plt.axis('off'); plt.show()

if sample_img and sample_img.lower().endswith('.dcm'):
    show_augments(sample_img)

# --- VISUAL SUMMARIES (HISTOGRAMS) --------------------------------------------
def hist_numeric(df, cols, title_prefix):
    for c in cols[:6]:  # limit to up to 6 histos to keep runtime reasonable
        plt.figure(figsize=(5,3))
        df[c].dropna().astype(float).plot(kind='hist', bins=20)
        plt.title(f"{title_prefix}: {c}")
        plt.xlabel(c); plt.ylabel('freq'); plt.tight_layout(); plt.show()

hist_numeric(clinical_df, clin_num_cols, "Histogram (clinical)")
hist_numeric(meta_df, meta_num_cols, "Histogram (metadata)")

# --- SIMPLE WORKFLOW DIAGRAM (matplotlib boxes/arrows) ------------------------
def flowchart():
    plt.figure(figsize=(8,6))
    ax = plt.gca(); ax.axis('off')

    boxes = {
        "ingest": (0.1, 0.8, 0.35, 0.12, "Ingest\n(GDrive → CSV/DICOM)"),
        "validate": (0.55, 0.8, 0.35, 0.12, "Validate Paths\n& Schema"),
        "eda": (0.1, 0.55, 0.35, 0.12, "EDA\n(dict, missing,\nduplicates, skew)"),
        "prep": (0.55, 0.55, 0.35, 0.12, "Prep\n(types, scaling,\noutliers/imbalance)"),
        "augment": (0.1, 0.3, 0.35, 0.12, "Image Augmentation\n(flips/rot/denoise)"),
        "model": (0.55, 0.3, 0.35, 0.12, "Modeling\n(train/valid/test)"),
        "monitor": (0.33, 0.05, 0.35, 0.12, "Monitoring &\nRevision Sheet")
    }

    for key,(x,y,w,h,label) in boxes.items():
        rect = plt.Rectangle((x,y), w,h, fill=False, linewidth=2)
        ax.add_patch(rect)
        ax.text(x+w/2, y+h/2, label, ha='center', va='center')

    def arrow(x1,y1,x2,y2):
        ax.annotate("", xy=(x2,y2), xytext=(x1,y1),
                    arrowprops=dict(arrowstyle="->", lw=2))

    arrow(0.45,0.86, 0.55,0.86)
    arrow(0.275,0.8, 0.275,0.67)
    arrow(0.725,0.8, 0.725,0.67)
    arrow(0.275,0.55, 0.55,0.55)
    arrow(0.275,0.3, 0.55,0.3)
    arrow(0.725,0.3, 0.5,0.17)

    plt.title("Bird's-eye Workflow Diagram")
    plt.tight_layout()
    png_path = out_dir/'workflow_diagram.png'
    plt.savefig(png_path, dpi=150)
    plt.show()
    print("Saved diagram to:", png_path)

flowchart()

# --- INTERPRETATION NOTES (print for report) ----------------------------------
def brief_interpretations():
    notes = []

    # Class distributions
    if 'vital_status' in clinical_df.columns:
        vs = clinical_df['vital_status'].value_counts().to_dict()
        notes.append(f"Vital status distribution: {vs}. Consider stratified splits / rebalancing if needed.")

    if len(num_clin) >= 2 and isinstance(C_clin, pd.DataFrame):
        strong = (C_clin.stack()
                  .dropna()
                  .loc[lambda s: (s.abs() >= 0.7) & (s.index.get_level_values(0)!=s.index.get_level_values(1))])
        notes.append(f"Clinical strong correlations (|r|>=0.7): {strong.to_dict()}")

    if len(num_meta) >= 2 and isinstance(C_meta, pd.DataFrame):
        strong_m = (C_meta.stack()
                    .dropna()
                    .loc[lambda s: (s.abs() >= 0.7) & (s.index.get_level_values(0)!=s.index.get_level_values(1))])
        notes.append(f"Metadata strong correlations (|r|>=0.7): {strong_m.to_dict()}")

    if notes:
        print("\nINTERPRETATION HIGHLIGHTS")
        for n in notes: print("-", n)

brief_interpretations()

print("\nAll artifacts (data dictionaries, diagram) saved to:", out_dir)
```

```python
# -----------------------------
# WEEK 6: EDA, FEATURE ENGINEERING & PREPROCESSING
# Paste after your loading cell (clinical_df, meta_df present)
# -----------------------------

# Install packages if needed (uncomment)
!pip install seaborn scipy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.feature_selection import VarianceThreshold

# ---------- Helper: show small tables neatly ----------
pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 120)

# ---------- 1) Quick data snapshot: shape, head, dtypes ----------
print("=== Snapshot: Clinical Data ===")
print("Shape:", clinical_df.shape)
display(clinical_df.head(4))

print("\n=== Snapshot: Metadata ===")
print("Shape:", meta_df.shape)
display(meta_df.head(4))

# ---------- 2) Feature description & types ----------
print("\n=== Feature types (clinical_df) ===")
dtypes = clinical_df.dtypes.reset_index()
dtypes.columns = ['feature','dtype']
display(dtypes)

# Build a simple feature table (example columns; expand as appropriate)
# Replace candidate names with actual columns in your clinical_df
CANDIDATE_FEATURES = [c for c in ['patient_id','age','sex','tumor_type','slice_thickness','survival_time'] if c in clinical_df.columns]
feat_table = []
for f in CANDIDATE_FEATURES:
    typ = 'categorical' if clinical_df[f].dtype == 'object' or clinical_df[f].dtype.name == 'category' else 'numeric'
    example_vals = clinical_df[f].dropna().unique()[:5].tolist()
    feat_table.append({'feature':f, 'type':typ, 'n_missing': clinical_df[f].isnull().sum(), 'examples': example_vals})
print("\n=== Feature dictionary (partial) ===")
display(pd.DataFrame(feat_table))

# ---------- 3) Descriptive stats (numeric) ----------
num_cols = clinical_df.select_dtypes(include=[np.number]).columns.tolist()
if num_cols:
    print("\n=== Descriptive statistics (numeric columns) ===")
    display(clinical_df[num_cols].describe().T)
else:
    print("\nNo numeric columns found in clinical_df for descriptive stats.")

# ---------- 4) Visualizations ----------
# 4a: Histograms for numeric columns
if num_cols:
    clinical_df[num_cols].hist(bins=15, figsize=(12, 3*len(num_cols)))
    plt.suptitle("Numeric Feature Histograms")
    plt.show()

# 4b: Categorical counts (top categorical cols)
cat_cols = clinical_df.select_dtypes(include=['object','category','bool']).columns.tolist()
if cat_cols:
    top_cat = cat_cols[:3]  # show up to first 3 categorical features
    for c in top_cat:
        plt.figure(figsize=(6,3))
        clinical_df[c].value_counts(dropna=False).plot(kind='bar')
        plt.title(f"Counts: {c}")
        plt.xlabel(c); plt.ylabel("count")
        plt.show()

# 4c: Correlation heatmap (numeric)
if len(num_cols) >= 2:
    plt.figure(figsize=(8,6))
    sns.heatmap(clinical_df[num_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
    plt.title("Numeric Feature Correlations")
    plt.show()
else:
    print("Not enough numeric columns for correlation heatmap.")

# 4d: Scatter / relationship example: age vs survival_time (if both exist)
if 'age' in clinical_df.columns and 'survival_time' in clinical_df.columns:
    plt.figure(figsize=(6,4))
    sns.scatterplot(data=clinical_df, x='age', y='survival_time')
    plt.title("Age vs Survival Time")
    plt.show()
    # Simple Pearson correlation and p-value
    corr, pval = stats.pearsonr(clinical_df['age'].fillna(0), clinical_df['survival_time'].fillna(0))
    print(f"Pearson corr (age, survival_time): {corr:.3f}, p-value={pval:.3e}")

# ---------- 5) Inferential/stat test examples ----------
# Example: check if age differs by tumor_type (ANOVA) if tumor_type exists
if 'age' in clinical_df.columns and 'tumor_type' in clinical_df.columns:
    groups = [vals['age'].dropna().values for _, vals in clinical_df.groupby('tumor_type')]
    if len(groups) > 1:
        fstat, p = stats.f_oneway(*groups)
        print(f"\nANOVA age ~ tumor_type: F={fstat:.3f}, p={p:.3e}")
        if p < 0.05:
            print("=> Suggests significant age differences across tumor types (p<0.05).")
        else:
            print("=> No significant age differences across tumor types (p>=0.05).")

# ---------- 6) Interpretations / Insights (printed) ----------
print("\n=== Preliminary Interpretations & Insights ===")
# Example heuristics; tailor to real computed stats
if 'age' in clinical_df.columns:
    median_age = clinical_df['age'].median()
    print(f"- Median patient age: {median_age}")
if 'tumor_type' in clinical_df.columns:
    counts = clinical_df['tumor_type'].value_counts()
    print(f"- Tumor type distribution (top 3):\n{counts.head(3).to_string()}")
if num_cols:
    print("- Check correlation heatmap to identify strongly correlated features to consider for elimination or combination.")
print("- Missing value counts printed above — columns with many missing values may require imputation or removal.")
print("- If tumor classes are imbalanced, plan augmentations or re-sampling for modeling.")

# ---------- 7) Preprocessing: imputation, encoding, scaling (example pipeline) ----------
# We'll build a small reusable preprocessing snippet for tabular clinical features.

def preprocess_clinical(df, features, numeric_strategy='median', scale=True, onehot=True):
    """
    Simple preprocessing for tabular clinical features:
      - imputes numeric with median, categorical with mode
      - optionally applies standard scaling to numeric features
      - optionally one-hot encodes categorical features
    Returns: processed_df, fitted_imputers/scalers (dictionary)
    """
    df = df.copy()
    num_feats = [c for c in features if c in df.columns and pd.api.types.is_numeric_dtype(df[c])]
    cat_feats = [c for c in features if c in df.columns and not pd.api.types.is_numeric_dtype(df[c])]

    # Impute numeric
    num_imp = SimpleImputer(strategy=numeric_strategy)
    if num_feats:
        df[num_feats] = num_imp.fit_transform(df[num_feats])

    # Impute categorical
    cat_imp = SimpleImputer(strategy='most_frequent')
    if cat_feats:
        df[cat_feats] = cat_imp.fit_transform(df[cat_feats].astype(str))

    # One-hot encode categorical if requested
    enc = None
    if onehot and cat_feats:
        enc = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
        encoded = enc.fit_transform(df[cat_feats])
        enc_cols = enc.get_feature_names_out(cat_feats)
        df_enc = pd.DataFrame(encoded, columns=enc_cols, index=df.index)
    else:
        df_enc = df[cat_feats] if cat_feats else pd.DataFrame(index=df.index)

    # Scale numeric
    scaler = None
    if scale and num_feats:
        scaler = StandardScaler()
        scaled = scaler.fit_transform(df[num_feats])
        df_scaled = pd.DataFrame(scaled, columns=[f"{c}_z" for c in num_feats], index=df.index)
    else:
        df_scaled = df[num_feats] if num_feats else pd.DataFrame(index=df.index)

    # Combine final processed df
    processed = pd.concat([df_scaled, df_enc], axis=1)
    return processed, {'num_imputer': num_imp, 'cat_imputer': cat_imp, 'onehot': enc, 'scaler': scaler}

# Choose features to include in modeling (example selection)
# Start from candidate features earlier; you may refine based on domain knowledge
selected_features = CANDIDATE_FEATURES.copy()
print("\nSelected features for preprocessing:", selected_features)

processed_tabular, pipeline_objs = preprocess_clinical(clinical_df, selected_features, scale=True, onehot=True)
print("Processed tabular shape:", processed_tabular.shape)
display(processed_tabular.head(3))

# ---------- 8) Feature engineering: selection / construction / elimination ----------
print("\n=== Feature Engineering ===")
# 8a: Feature selection via low-variance filter (drop nearly-constant features)
vt = VarianceThreshold(threshold=1e-3)  # remove near-constant features
if processed_tabular.shape[1] > 0:
    vt.fit(processed_tabular.fillna(0))
    keep_mask = vt.get_support()
    kept_cols = processed_tabular.columns[keep_mask].tolist()
    dropped_cols = [c for c in processed_tabular.columns if c not in kept_cols]
    print("Dropped low-variance columns:", dropped_cols)
    processed_tabular = processed_tabular[kept_cols]
else:
    print("No tabular columns to filter for variance.")

# 8b: Feature construction example: age buckets (if age exists)
if 'age' in clinical_df.columns:
    clinical_df['age_bucket'] = pd.cut(clinical_df['age'], bins=[0,30,50,70,120], labels=['<30','30-50','50-70','70+'])
    # encode this new feature and add to processed_tabular
    ab_enc = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
    ab_arr = ab_enc.fit_transform(clinical_df[['age_bucket']].astype(str))
    ab_cols = ab_enc.get_feature_names_out(['age_bucket'])
    ab_df = pd.DataFrame(ab_arr, columns=ab_cols, index=clinical_df.index)
    processed_tabular = pd.concat([processed_tabular, ab_df], axis=1)
    print("Constructed feature: age_bucket -> one-hot added.")

# 8c: Feature elimination (example): drop features with >50% missing originally
missing_frac = clinical_df.isnull().mean()
to_drop = missing_frac[missing_frac > 0.5].index.tolist()
if to_drop:
    print("Eliminating features with >50% missing:", to_drop)
    processed_tabular = processed_tabular.drop(columns=[c for c in to_drop if c in processed_tabular.columns], errors='ignore')
else:
    print("No features with >50% missing.")

display(processed_tabular.head(3))

# ---------- 9) Scaling / normalization notes ----------
print("\n=== Scaling/Normalization Recommendation ===")
print("- For neural networks (CNN/MLP), standardized numeric features (zero mean, unit variance) are recommended.")
print("- For tree-based models (RandomForest, XGBoost), scaling is not strictly required.")
print("- We applied StandardScaler to numeric features and OneHot encoding to categoricals in the processed tabular output.")

# ---------- 10) Visualize final processed feature distributions (a few examples) ----------
if processed_tabular.shape[1] > 0:
    # show first up to 6 features distributions
    cols_to_plot = processed_tabular.columns[:6].tolist()
    processed_tabular[cols_to_plot].hist(bins=20, figsize=(12, 2.5*len(cols_to_plot)))
    plt.suptitle("Processed feature distributions (first few)")
    plt.show()

# ---------- 11) Save processed features for modeling ----------
##out_path = Path('/content/drive/My Drive/capstone/tumor_pred_ai/DataSet/preprocessed/clinical_features_model_v1.csv')
##out_path.parent.mkdir(parents=True, exist_ok=True)
##processed_tabular.to_csv(out_path, index=False)
##print("Saved processed tabular features to:", out_path)

# ---------- 12) Final short interpretation (print) ----------
print("\n=== Final Interpretation Summary ===")
print("- We completed descriptive and simple inferential checks. See above plots for distributions and correlations.")
print("- Key insights: check tumor_type distribution for imbalance (shown above), age distribution, and any significant correlations.")
print("- Preprocessing performed: imputation (median/mode), standard scaling, one-hot encoding, low-variance filtering, simple constructed feature (age_bucket).")
print("- These processed features are ready to feed into an MLP (mlp_clinical_v1) or to be concatenated with image embeddings in a fusion model (fusion_cnn_mlp_v1).")

# End of cell
```

```python
# ========================== IMAGE EDA (Appendix D) ============================
# Saves: normalized slice, HU histogram, CLAHE before/after, series HU histogram,
# slice-wise HU mean/std profiles, montage, spacing/thickness histograms and CSVs.

import os, numpy as np, pandas as pd, matplotlib.pyplot as plt, pydicom
from pathlib import Path

# Output folder for appendix images
img_out = (out_dir / "image_eda")
img_out.mkdir(parents=True, exist_ok=True)

def save_fig(path):
    plt.tight_layout()
    plt.savefig(path, dpi=200, bbox_inches="tight")
    plt.close()

def dcm_to_hu(ds):
    arr = ds.pixel_array.astype(np.float32)
    slope = float(getattr(ds, "RescaleSlope", 1.0))
    intercept = float(getattr(ds, "RescaleIntercept", 0.0))
    return arr * slope + intercept

def hu_window(arr, hu_min=-150.0, hu_max=250.0):
    arr = np.clip(arr, hu_min, hu_max)
    return ((arr - hu_min) / max(hu_max - hu_min, 1e-6)).astype(np.float32)

def pick_series_dir_from_dicoms(dicoms_list):
    # Prefer non-segmentation, non-RTSTRUCT paths (fast filter by path string)
    prefer = [p for p in dicoms_list if ("Segmentation" not in p and "RTSTRUCT" not in p and "rtstruct" not in p)]
    cand = prefer if len(prefer) > 0 else dicoms_list
    if len(cand) == 0:
        raise FileNotFoundError("No DICOM files available.")
    return os.path.dirname(cand[0])

def list_series_files(series_dir):
    # Collect files in the chosen series directory only (no recursion)
    files = [os.path.join(series_dir, f) for f in os.listdir(series_dir)
             if os.path.isfile(os.path.join(series_dir, f))]
    # Prefer .dcm first; if none, keep all and we will try reading them
    dcm_only = [p for p in files if p.lower().endswith(".dcm")]
    files = dcm_only if len(dcm_only) > 0 else files
    files = sorted(files)
    return files

def sort_by_instance_number(paths):
    # Lightweight header read to sort by InstanceNumber for correct ordering
    def inst_num(p):
        try:
            ds = pydicom.dcmread(p, stop_before_pixels=True, force=True)
            return int(getattr(ds, "InstanceNumber", 10**9))
        except Exception:
            return 10**9
    # For speed, only sample up to 120 files when measuring instance numbers
    if len(paths) > 120:
        idx = np.linspace(0, len(paths)-1, 120).astype(int)
        sample = [paths[i] for i in idx]
        sample_pairs = [(p, inst_num(p)) for p in sample]
        # If instance numbers look useful, map back approximate order
        if not all(v == 10**9 for _, v in sample_pairs):
            order = np.argsort([inst_num(p) for p in paths])
            return [paths[i] for i in order]
        else:
            return paths  # fallback to filename sort already applied
    else:
        pairs = [(p, inst_num(p)) for p in paths]
        if all(v == 10**9 for _, v in pairs):
            return paths
        return [p for p,_ in sorted(pairs, key=lambda t: t[1])]

# 1) Choose a single CT series quickly and robustly
series_dir = pick_series_dir_from_dicoms(dicoms)
series_files = list_series_files(series_dir)
if len(series_files) == 0:
    raise FileNotFoundError(f"No files in series directory: {series_dir}")

series_files = sort_by_instance_number(series_files)
mid_idx = max(0, len(series_files)//2 - 1)
sample_dcm = series_files[mid_idx]
print(f"[Image EDA] Using series_dir: {series_dir}")
print(f"[Image EDA] Files in series: {len(series_files)}")
print(f"[Image EDA] Middle slice: {sample_dcm}")

# 2) One-slice EDA: normalized window, HU histogram, CLAHE before/after
ds_mid = pydicom.dcmread(sample_dcm)
hu_mid = dcm_to_hu(ds_mid)
win_mid = hu_window(hu_mid, -150.0, 250.0)

plt.figure(figsize=(4,4)); plt.imshow(win_mid, cmap="gray"); plt.title("Normalized CT slice"); plt.axis("off")
save_fig(img_out / "normalized_ct_slice.png")

plt.figure(figsize=(5,3)); plt.hist(hu_mid[np.isfinite(hu_mid)].ravel(), bins=128)
plt.title("HU intensity histogram (slice)"); plt.xlabel("HU"); plt.ylabel("count")
save_fig(img_out / "hu_histogram.png")

try:
    import cv2
except ImportError:
    import sys, subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "opencv-python-headless", "-q"])
    import cv2

u8 = np.clip(win_mid * 255.0, 0, 255).astype(np.uint8)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
cla = clahe.apply(u8).astype(np.float32) / 255.0

plt.figure(figsize=(4,4)); plt.imshow(win_mid, cmap="gray"); plt.title("Before CLAHE"); plt.axis("off")
save_fig(img_out / "before_clahe.png")
plt.figure(figsize=(4,4)); plt.imshow(cla, cmap="gray"); plt.title("After CLAHE"); plt.axis("off")
save_fig(img_out / "after_clahe.png")

print("PixelSpacing:", getattr(ds_mid, "PixelSpacing", None),
      "SliceThickness:", getattr(ds_mid, "SliceThickness", None))

# 3) Series-level EDA: sample up to 48 slices evenly for speed
S_MAX = min(48, len(series_files))
sel_idx = np.linspace(0, len(series_files)-1, S_MAX).astype(int)
stack = []
pixel_spacing = []
slice_thickness = []

for i in sel_idx:
    try:
        ds = pydicom.dcmread(series_files[i])
        hu = dcm_to_hu(ds)
        stack.append(hu)
        ps = getattr(ds, "PixelSpacing", None)
        if ps and len(ps) >= 1:
            try:
                pixel_spacing.append(float(ps[0]))
            except Exception:
                pass
        th = getattr(ds, "SliceThickness", None)
        if th is not None:
            try:
                slice_thickness.append(float(th))
            except Exception:
                pass
    except Exception:
        pass

if len(stack) >= 2:
    vol = np.stack(stack, axis=0)  # (S, H, W)
    S, H, W = vol.shape
    # Series HU histogram
    plt.figure(figsize=(5,3))
    plt.hist(vol[np.isfinite(vol)].ravel(), bins=256)
    plt.title("HU intensity histogram (series sample)")
    plt.xlabel("HU"); plt.ylabel("count")
    save_fig(img_out / "series_hu_histogram.png")

    # Slice mean and std profiles
    slice_means = vol.reshape(S, -1).mean(axis=1)
    slice_stds  = vol.reshape(S, -1).std(axis=1)

    plt.figure(figsize=(6,3))
    plt.plot(np.arange(S), slice_means)
    plt.title("Slice mean HU across series"); plt.xlabel("slice index"); plt.ylabel("mean HU")
    save_fig(img_out / "series_slice_meanHU.png")

    plt.figure(figsize=(6,3))
    plt.plot(np.arange(S), slice_stds)
    plt.title("Slice std HU across series"); plt.xlabel("slice index"); plt.ylabel("std HU")
    save_fig(img_out / "series_slice_stdHU.png")

    # Save numeric summary CSV
    pd.DataFrame({"slice_index": np.arange(S),
                  "mean_hu": slice_means,
                  "std_hu": slice_stds}).to_csv(img_out / "series_slice_hu_summary.csv", index=False)

    # Montage of evenly spaced slices (windowed)
    def window01(a, lo=-150.0, hi=250.0):
        b = np.clip(a, lo, hi)
        return ((b - lo) / max(hi - lo, 1e-6)).astype(np.float32)

    win_vol = np.stack([window01(vol[i]) for i in range(S)], axis=0)

    def save_montage(win_vol, rows=4, cols=12, out_name="series_montage.png"):
        total = rows * cols
        k = min(total, win_vol.shape[0])
        picks = np.linspace(0, win_vol.shape[0]-1, k).astype(int)
        plt.figure(figsize=(cols*1.6, rows*1.6))
        for j, i in enumerate(picks):
            plt.subplot(rows, cols, j+1)
            plt.imshow(win_vol[i], cmap="gray")
            plt.axis("off")
        save_fig(img_out / out_name)

    save_montage(win_vol, rows=4, cols=12, out_name="series_montage.png")

# 4) Geometry histograms and CSVs
if len(slice_thickness) > 0:
    plt.figure(figsize=(5,3))
    plt.hist(slice_thickness, bins=20)
    plt.title("Slice thickness distribution (mm)")
    plt.xlabel("mm"); plt.ylabel("count")
    save_fig(img_out / "series_slice_thickness_hist.png")
    pd.DataFrame({"slice_thickness_mm": slice_thickness}).to_csv(img_out / "series_slice_thickness.csv", index=False)

if len(pixel_spacing) > 0:
    plt.figure(figsize=(5,3))
    plt.hist(pixel_spacing, bins=20)
    plt.title("Pixel spacing distribution (row, mm)")
    plt.xlabel("mm"); plt.ylabel("count")
    save_fig(img_out / "series_pixel_spacing_hist.png")
    pd.DataFrame({"pixel_spacing_row_mm": pixel_spacing}).to_csv(img_out / "series_pixel_spacing.csv", index=False)

print("Image EDA artifacts saved to:", img_out)
# ==============================================================================

```

```python
# EDA image analysis for GPU support to process large Image data Set
# ======================= IMAGE EDA at SCALE (7k series) ======================
# Parallel, fast-decoder, local-disk optimized. Produces:
#  - Per-series: normalized slice, HU histogram, CLAHE before/after, montage, HU mean/std
#  - Global CSV: one row per series with summary metrics
# Notes:
#  - GPU (A100) isn't the main accelerator here; I/O is bottleneck. We focus on parallel I/O + vectorized NumPy.
#  - Use 8 workers; keep BLAS threads at 1 to avoid oversubscription.

import os, sys, time, re, shutil, numpy as np, pandas as pd, matplotlib.pyplot as plt
from pathlib import Path

# ---- Speed-ups (fast decoders + thread caps) ----
!pip -q install pylibjpeg pylibjpeg-libjpeg pylibjpeg-openjpeg gdcm opencv-python-headless joblib tqdm
import pydicom
pydicom.config.use_none_as_empty_text = True  # small skip on bad tags

# Cap BLAS threads so 8-process parallel doesn't oversubscribe CPU cores
os.environ["OPENBLAS_NUM_THREADS"]  = "1"
os.environ["MKL_NUM_THREADS"]       = "1"
os.environ["NUMEXPR_MAX_THREADS"]   = "1"
os.environ["OMP_NUM_THREADS"]       = "1"

from joblib import Parallel, delayed
from tqdm import tqdm
import pydicom
import cv2

# ---------- Optional: stage data to local SSD (MUCH faster than Drive) ----------
#  for many cases, uncomment to copy once. Adjust source as needed.
# !rsync -a "/content/drive/My Drive/capstone/tumor_pred_ai/DataSet/manifest-1592488683281/C4KC-KiTS/" /content/kits_local/
# IMAGE_ROOT = Path("/content/kits_local")  # switch to local staging for the big run

# If you didn't rsync, keep your existing IMAGE_ROOT from above.
assert IMAGE_ROOT.exists(), f"IMAGE_ROOT not found: {IMAGE_ROOT}"

# ---------- Outputs ----------
img_root = out_dir / "image_eda_bulk"
img_root.mkdir(parents=True, exist_ok=True)
summary_csv = img_root / "series_summary_metrics.csv"

# ---------- Helpers ----------
def save_fig(path):
    plt.tight_layout()
    plt.savefig(path, dpi=200, bbox_inches="tight")
    plt.close()

def dcm_to_hu(ds):
    arr = ds.pixel_array.astype(np.float32)
    slope = float(getattr(ds, "RescaleSlope", 1.0))
    intercept = float(getattr(ds, "RescaleIntercept", 0.0))
    return arr * slope + intercept

def hu_window(arr, hu_min=-150.0, hu_max=250.0):
    arr = np.clip(arr, hu_min, hu_max)
    return ((arr - hu_min) / max(hu_max - hu_min, 1e-6)).astype(np.float32)

def list_series_dirs(root: Path):
    """Yield leaf dirs that contain at least one DICOM; skip obvious segmentation/RTSTRUCT."""
    out = []
    for dp, dn, fn in os.walk(str(root)):
        if not fn:
            continue
        if ("Segmentation" in dp) or ("RTSTRUCT" in dp) or ("rtstruct" in dp):
            continue
        if any(f.lower().endswith(".dcm") for f in fn):
            out.append(dp)
    return out

def list_series_files(series_dir):
    files = [os.path.join(series_dir, f) for f in os.listdir(series_dir)
             if os.path.isfile(os.path.join(series_dir, f))]
    dcm = [p for p in files if p.lower().endswith(".dcm")]
    return sorted(dcm if dcm else files)

def sort_by_instance_number(paths):
    def inst_num(p):
        try:
            ds = pydicom.dcmread(p, stop_before_pixels=True, force=True)
            return int(getattr(ds, "InstanceNumber", 10**9))
        except Exception:
            return 10**9
    # For long series, avoid header storm by sampling to decide usefulness
    if len(paths) > 120:
        sample_idx = np.linspace(0, len(paths)-1, 120).astype(int)
        if not all(inst_num(paths[i]) == 10**9 for i in sample_idx):
            order = np.argsort([inst_num(p) for p in paths])
            return [paths[i] for i in order]
        return paths
    pairs = [(p, inst_num(p)) for p in paths]
    if all(v == 10**9 for _, v in pairs):
        return paths
    return [p for p,_ in sorted(pairs, key=lambda t: t[1])]

def safe_series_name(path_str: str):
    # stable short name for folder (patient/series tail)
    tail = path_str.strip("/").split("/")[-2:]
    base = "_".join(tail)
    base = re.sub(r"[^A-Za-z0-9._-]+", "_", base)
    return base[:120]

# ---------- Per-series worker ----------
def process_one_series(series_dir: str, S_MAX=16):
    """Process one series; save artifacts under its own subfolder; return summary metrics."""
    try:
        files = list_series_files(series_dir)
        if not files:
            return {"series_dir": series_dir, "ok": False, "reason": "no files"}

        files = sort_by_instance_number(files)
        # Pick middle slice for quick visuals
        mid_idx = max(0, len(files)//2 - 1)
        mid_path = files[mid_idx]

        # Where to save this series' artifacts
        sub = img_root / safe_series_name(series_dir)
        sub.mkdir(exist_ok=True, parents=True)

        # Read mid slice
        ds_mid = pydicom.dcmread(mid_path)
        hu_mid = dcm_to_hu(ds_mid)
        win_mid = hu_window(hu_mid, -150.0, 250.0)

        # Save normalized slice, histogram, CLAHE before/after
        plt.figure(figsize=(4,4)); plt.imshow(win_mid, cmap="gray"); plt.title("Normalized CT slice"); plt.axis("off")
        save_fig(sub / "normalized_ct_slice.png")

        plt.figure(figsize=(5,3)); plt.hist(hu_mid[np.isfinite(hu_mid)].ravel(), bins=128)
        plt.title("HU intensity histogram (slice)"); plt.xlabel("HU"); plt.ylabel("count")
        save_fig(sub / "hu_histogram.png")

        u8 = np.clip(win_mid * 255.0, 0, 255).astype(np.uint8)
        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
        cla  = clahe.apply(u8).astype(np.float32) / 255.0
        plt.figure(figsize=(4,4)); plt.imshow(win_mid, cmap="gray"); plt.title("Before CLAHE"); plt.axis("off")
        save_fig(sub / "before_clahe.png")
        plt.figure(figsize=(4,4)); plt.imshow(cla, cmap="gray"); plt.title("After CLAHE"); plt.axis("off")
        save_fig(sub / "after_clahe.png")

        pix_sp = getattr(ds_mid, "PixelSpacing", None)
        sl_th  = getattr(ds_mid, "SliceThickness", None)
        px_row = float(pix_sp[0]) if (pix_sp and len(pix_sp)>0) else np.nan
        thickness_mid = float(sl_th) if sl_th is not None else np.nan

        # Sample up to S_MAX evenly for series metrics and montage
        S_MAX = min(S_MAX, len(files))
        idx = np.linspace(0, len(files)-1, S_MAX).astype(int)
        stack = []
        px_all = []
        th_all = []
        for i in idx:
            try:
                ds = pydicom.dcmread(files[i])
                hu = dcm_to_hu(ds)
                stack.append(hu)
                ps = getattr(ds, "PixelSpacing", None)
                if ps and len(ps)>=1:
                    try: px_all.append(float(ps[0]))
                    except: pass
                th = getattr(ds, "SliceThickness", None)
                if th is not None:
                    try: th_all.append(float(th))
                    except: pass
            except Exception:
                pass

        series_hu_mean = np.nan
        series_hu_std  = np.nan
        if len(stack) >= 2:
            vol = np.stack(stack, axis=0)  # (S, H, W)
            S, H, W = vol.shape

            # Global HU histogram for series sample
            plt.figure(figsize=(5,3))
            plt.hist(vol[np.isfinite(vol)].ravel(), bins=256)
            plt.title("HU intensity histogram (series sample)")
            plt.xlabel("HU"); plt.ylabel("count")
            save_fig(sub / "series_hu_histogram.png")

            # Slice mean/std profiles
            s_means = vol.reshape(S, -1).mean(axis=1)
            s_stds  = vol.reshape(S, -1).std(axis=1)
            series_hu_mean = float(np.nanmean(s_means))
            series_hu_std  = float(np.nanmean(s_stds))

            plt.figure(figsize=(6,3)); plt.plot(np.arange(S), s_means)
            plt.title("Slice mean HU across series"); plt.xlabel("slice index"); plt.ylabel("mean HU")
            save_fig(sub / "series_slice_meanHU.png")
            plt.figure(figsize=(6,3)); plt.plot(np.arange(S), s_stds)
            plt.title("Slice std HU across series"); plt.xlabel("slice index"); plt.ylabel("std HU")
            save_fig(sub / "series_slice_stdHU.png")

            pd.DataFrame({"slice_index": np.arange(S),
                          "mean_hu": s_means, "std_hu": s_stds}).to_csv(sub / "series_slice_hu_summary.csv", index=False)

            # Montage
            def window01(a, lo=-150.0, hi=250.0):
                b = np.clip(a, lo, hi)
                return ((b - lo) / max(hi - lo, 1e-6)).astype(np.float32)
            win_vol = np.stack([window01(vol[i]) for i in range(S)], axis=0)

            rows, cols = 4, 12
            total = rows * cols
            k = min(total, win_vol.shape[0])
            picks = np.linspace(0, win_vol.shape[0]-1, k).astype(int)
            plt.figure(figsize=(cols*1.6, rows*1.6))
            for j, ii in enumerate(picks):
                plt.subplot(rows, cols, j+1)
                plt.imshow(win_vol[ii], cmap="gray")
                plt.axis("off")
            save_fig(sub / "series_montage.png")

        # Geometry quick-hists (per series; optional – most useful globally)
        if th_all:
            plt.figure(figsize=(5,3)); plt.hist(th_all, bins=20)
            plt.title("Slice thickness distribution (mm)"); plt.xlabel("mm"); plt.ylabel("count")
            save_fig(sub / "series_slice_thickness_hist.png")
            pd.DataFrame({"slice_thickness_mm": th_all}).to_csv(sub / "series_slice_thickness.csv", index=False)

        if px_all:
            plt.figure(figsize=(5,3)); plt.hist(px_all, bins=20)
            plt.title("Pixel spacing distribution (row, mm)"); plt.xlabel("mm"); plt.ylabel("count")
            save_fig(sub / "series_pixel_spacing_hist.png")
            pd.DataFrame({"pixel_spacing_row_mm": px_all}).to_csv(sub / "series_pixel_spacing.csv", index=False)

        return {
            "series_dir": series_dir,
            "folder": str(sub),
            "n_files": len(files),
            "sampled_slices": int(S_MAX),
            "pixel_spacing_row_mid": float(px_row) if np.isfinite(px_row) else np.nan,
            "slice_thickness_mid": float(thickness_mid) if np.isfinite(thickness_mid) else np.nan,
            "series_meanHU_mean": series_hu_mean,
            "series_meanHU_std": series_hu_std,
            "ok": True
        }
    except Exception as e:
        return {"series_dir": series_dir, "ok": False, "reason": str(e)}

# ---------- Collect series and run parallel with 8 workers ----------
series_dirs = list_series_dirs(IMAGE_ROOT)
print("Total candidate series:", len(series_dirs))

#  ~7000+, To test first, limit like: series_dirs = series_dirs[:200]
N_WORKERS = 8             # requested 8 threads (processes)
S_MAX     = 16            # faster than 48; good for QC + appendix (adjust if needed)

t0 = time.time()
results = Parallel(n_jobs=N_WORKERS, backend="loky", prefer="processes")(
    delayed(process_one_series)(sd, S_MAX=S_MAX) for sd in tqdm(series_dirs, desc="Image EDA")
)
dt = time.time() - t0

# ---------- Save global summary ----------
summary = pd.DataFrame(results)
summary.to_csv(summary_csv, index=False)

ok = int(summary["ok"].sum())
bad = int((~summary["ok"]).sum())
print(f"Done: {ok} ok, {bad} failed in {dt/60:.1f} min with {N_WORKERS} workers.")
print("Global metrics CSV:", summary_csv)
print("Per-series artifacts under:", img_root)
# ==============================================================================
```

```python
# ========================== FEATURE SELECTION (Appendix E) ==========================
# Generates: feature_selection_scores.csv and feature_elimination_log.csv

import os
from sklearn.feature_selection import mutual_info_classif, f_classif, VarianceThreshold
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

# Paths
feat_out = out_dir / "feature_engineering"
feat_out.mkdir(parents=True, exist_ok=True)

# Ensure target column exists in your clinical dataset
TARGET_COL = "vital_status"  # <-- change if your target is named differently

# Select only numeric columns
X = clinical_df.select_dtypes(include=[np.number]).copy()
if TARGET_COL in clinical_df.columns:
    y = clinical_df[TARGET_COL]
else:
    raise ValueError(f"Target column '{TARGET_COL}' not found in clinical_df")

# Drop columns that are all NaN before imputation
X = X.dropna(axis=1, how='all')

# Handle missing values (impute with median)
X = X.fillna(X.median())

# Standardize numeric features
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)

# --- Mutual Information and ANOVA F-score ---
# Need to encode the target variable for mutual_info_classif and f_classif
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_encoded = le.fit_transform(y)

mi_scores = mutual_info_classif(X_scaled, y_encoded, random_state=42)
f_scores, _ = f_classif(X_scaled, y_encoded)

scores_df = pd.DataFrame({
    "feature": X.columns,
    "mutual_info": mi_scores,
    "anova_f": f_scores
}).sort_values(by="mutual_info", ascending=False)

scores_df.to_csv(feat_out / "feature_selection_scores.csv", index=False)

# --- Variance Threshold Filtering ---
selector = VarianceThreshold(threshold=0.01)  # Increased threshold slightly
selector.fit(X_scaled)
kept_features = X.columns[selector.get_support()].tolist()
removed_low_variance = set(X.columns) - set(kept_features)

# --- Correlation Pruning ---
corr_matrix = X_scaled.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
high_corr = [column for column in upper.columns if any(upper[column] > 0.9)]

removed_corr = set(high_corr)
removed_total = removed_low_variance.union(removed_corr)

elim_log = pd.DataFrame({
    "removed_feature": list(removed_total),
    "reason": ["low_variance" if f in removed_low_variance else "high_corr"
               for f in removed_total]
})
elim_log.to_csv(feat_out / "feature_elimination_log.csv", index=False)

print(f" Feature selection completed.")
print(f"  -> Saved scores to: {feat_out/'feature_selection_scores.csv'}")
print(f"  -> Saved elimination log to: {feat_out/'feature_elimination_log.csv'}")
print(f"Remaining features: {len(X.columns) - len(removed_total)} / {len(X.columns)}")

# Optional: visualize top features
plt.figure(figsize=(6,4))
top = scores_df.head(15)
plt.barh(top['feature'], top['mutual_info'], color='gray')
plt.gca().invert_yaxis()
plt.title("Top 15 Features by Mutual Information")
plt.xlabel("Mutual Information")
plt.tight_layout()
plt.show()
# ==============================================================================
```

```python
# pip install pandas numpy scikit-learn seaborn matplotlib

import numpy as np
import pandas as pd
from sklearn.feature_selection import (
    mutual_info_classif, mutual_info_regression,
    f_classif, f_regression
)
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
import matplotlib.pyplot as plt
import seaborn as sns

def compute_mi_anova(
    df: pd.DataFrame,
    target: str,
    task: str = "classification",  # "classification" or "regression"
    top_k: int | None = None
) -> pd.DataFrame:
    """
    Returns a DataFrame with MI and ANOVA scores (and ranks) for all predictors vs target.
    Automatically one-hot encodes categoricals, imputes missing values, and aggregates
    one-hot blocks back to the original feature using max score.
    Handles columns that are all NaN by dropping them before preprocessing.
    """
    assert task in {"classification", "regression"}
    y = df[target].values
    X = df.drop(columns=[target])

    # Drop columns that are all NaN before preprocessing
    X = X.dropna(axis=1, how='all')

    numeric_cols = X.select_dtypes(include=[np.number]).columns.tolist()
    categorical_cols = X.select_dtypes(exclude=[np.number]).columns.tolist()

    # Preprocess: impute numerics (median) and categoricals (most_frequent), then one-hot
    # Use remainder='passthrough' to keep other columns if any, though we dropped all-NaN
    pre = ColumnTransformer(
        transformers=[
            ("num", SimpleImputer(strategy="median"), numeric_cols),
            ("cat", Pipeline([
                ("imp", SimpleImputer(strategy="most_frequent")),
                ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
            ]), categorical_cols),
        ],
        remainder="passthrough" # Keep any other columns
    )

    Xp = pre.fit_transform(X)

    # Build names for transformed matrix
    feature_names = []
    # Get names from the preprocessor's transformations
    for name, transformer, cols in pre.transformers_:
        if transformer != 'drop' and hasattr(transformer, 'get_feature_names_out'):
             feature_names.extend(transformer.get_feature_names_out(cols).tolist())
        elif cols: # handle cases like SimpleImputer which don't have get_feature_names_out but keep original names
            feature_names.extend(cols)
    # Add names from remainder (passthrough) if any
    if pre.remainder == 'passthrough':
        remainder_cols = pre.get_feature_names_out()
        # filter out names from transformers
        feature_names.extend([col for col in remainder_cols if col not in feature_names])


    # Choose scorers
    if task == "classification":
        # Need to encode the target variable for mutual_info_classif and f_classif
        from sklearn.preprocessing import LabelEncoder
        le = LabelEncoder()
        y_encoded = le.fit_transform(y)
        mi = mutual_info_classif(Xp, y_encoded, discrete_features=False, random_state=0)
        fvals, _ = f_classif(Xp, y_encoded)
    else:
        mi = mutual_info_regression(Xp, y, discrete_features=False, random_state=0)
        fvals, _ = f_regression(Xp, y)

    # Add a check to ensure lengths match
    if not (len(feature_names) == len(mi) == len(fvals)):
        print(f"Warning: Length mismatch. Feature names: {len(feature_names)}, MI scores: {len(mi)}, ANOVA scores: {len(fvals)}")
        # Attempt to align based on the shorter list, or raise error
        min_len = min(len(feature_names), len(mi), len(fvals))
        feature_names = feature_names[:min_len]
        mi = mi[:min_len]
        fvals = fvals[:min_len]
        # If still not matching, raise an error
        if not (len(feature_names) == len(mi) == len(fvals)):
             raise ValueError("Feature names and scores length mismatch after attempted alignment.")


    scores = pd.DataFrame({
        "feature_expanded": feature_names,
        "MI": mi,
        "ANOVA": fvals
    })

    # Collapse one-hot back to original feature by taking max score across its dummies
    def base_name(expanded):
        # numeric stay as-is; one-hot look like "col_catvalue"
        if expanded in numeric_cols:
            return expanded
        # For OHE names like "col__value" or "col_value" depending on version:
        # use split on first underscore after the original col name prefix
        for c in categorical_cols:
            if expanded.startswith(c + "_") or expanded.startswith(c + "__"):
                return c
        # Fallback for remainder columns if any
        if expanded in X.columns:
            return expanded
        return expanded # Should not happen with current logic

    scores["feature"] = scores["feature_expanded"].apply(base_name)
    agg = scores.groupby("feature", as_index=False).agg({"MI": "max", "ANOVA": "max"})

    # Create ranks (1 = strongest)
    agg["MI_rank"] = agg["MI"].rank(ascending=False, method="min")
    agg["ANOVA_rank"] = agg["ANOVA"].rank(ascending=False, method="min")

    # Optional: keep only top_k by either metric
    if top_k is not None:
        keep = agg.assign(best_rank=agg[["MI_rank", "ANOVA_rank"]].min(axis=1)) \
                  .nsmallest(top_k, "best_rank")["feature"]
        agg = agg[agg["feature"].isin(keep)].copy()

    # Normalize scores to [0,1] for comparable color scales in heatmaps
    for col in ["MI", "ANOVA"]:
        vmax = agg[col].max() or 1.0
        vmin = agg[col].min()
        rng = max(vmax - vmin, 1e-9)
        agg[col + "_norm"] = (agg[col] - vmin) / rng

    # Sort by average rank so plots look nice
    agg["avg_rank"] = (agg["MI_rank"] + agg["ANOVA_rank"]) / 2
    agg = agg.sort_values("avg_rank").reset_index(drop=True)
    return agg

def plot_mi_anova_heatmaps(agg: pd.DataFrame, title: str):
    """
    Two heatmaps (MI and ANOVA), same feature order, normalized 0-1 scales.
    """
    features = agg["feature"].tolist()

    if not features:
        print("No features to plot heatmaps for.")
        return

    fig, axes = plt.subplots(1, 2, figsize=(10.5, max(3, 0.4 * len(features))), constrained_layout=True)

    # MI heatmap
    mi_mat = agg[["MI_norm"]].T
    sns.heatmap(mi_mat, ax=axes[0], cmap="viridis", cbar=True, annot=False,
                xticklabels=features, yticklabels=["MI"], vmin=0.0, vmax=1.0)
    axes[0].set_title("Mutual Information (normalized)")
    axes[0].tick_params(axis='x', rotation=60) # Removed ha='right'

    # ANOVA heatmap
    anova_mat = agg[["ANOVA_norm"]].T
    sns.heatmap(anova_mat, ax=axes[1], cmap="magma", cbar=True, annot=False,
                xticklabels=features, yticklabels=["ANOVA F"], vmin=0.0, vmax=1.0)
    axes[1].set_title("ANOVA F-score (normalized)")
    axes[1].tick_params(axis='x', rotation=60) # Removed ha='right'

    fig.suptitle(title, fontsize=12, y=1.04)
    plt.show()

```

```python
# Generate MI and ANOVA scores and ranks for clinical data
agg_cls = compute_mi_anova(clinical_df, target="vital_status", task="classification", top_k=15)

# Plot heatmaps for normalized MI and ANOVA scores
plot_mi_anova_heatmaps(agg_cls, title="Top 15 Numerical Clinical Features vs Vital Status")

# Optional: Display the aggregated scores table
# display(agg_cls)
```

## Appendix E – Feature Engineering Evidence

This appendix documents the feature selection pipeline applied to **numerical clinical attributes**.

### Techniques Applied
- **Mutual Information (MI):** Captured nonlinear relevance between numerical features and the `vital_status` outcome.
- **ANOVA F-score:** Measured class-wise variance ratios to identify features with significantly different means across `vital_status` classes.
- **Variance Threshold Filtering:** Eliminated numerical features with very low variance (threshold < 0.01), which are unlikely to be predictive.
- **Correlation Pruning:** Identified and removed redundant numerical features with a high absolute correlation (|r| ≥ 0.9) to reduce multicollinearity.
- **Scaling and Normalization:** The final set of numerical features were standardized using `StandardScaler()` to have zero mean and unit variance, important for many machine learning models.

### Outputs and Artifacts
- `feature_selection_scores.csv` – Contains the calculated Mutual Information and ANOVA F-scores for each numerical clinical feature.
- `feature_elimination_log.csv` – Lists the numerical clinical features that were removed and the reason for elimination (low variance or high correlation).
- **Top Feature Visualization:** A bar plot showing the top numerical clinical features ranked by Mutual Information score.
